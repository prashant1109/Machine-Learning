**Now to clean the data set we need to fill the Null Values**
-- Now here we have two types of data Numerical and Categorical columns
1. The Categorical columns are filled by the MODE values i.e : We will the values with the occurance which has occured the most
2. The Numerical columns are filled by two types of data 
    2.1 : Mean : If We have continious values with no major outliers then we fill the values with mean
    2.2 : Median: If there are extreme values (Conatins outliers) then we fill the NA data with the Median 

**IMPORTANT**
If in a dataset if there are more than 50% values as NAN then we should exclude that columns from the datasets 
if there are Less than 50% values as NAN then we should try to impute (Fill) those values
If there is a dataset where there are few few null values then in those cases we can drop those rows itself

### Now Lets fill the NA data for Numerical Column
**Important Rules** 
1. Here since we have outliers present and this column is a numerical coulun , we will will the data with Median values
2. If we want to fill the data in the originaly loaded dataframe itself then we will use inplace = True --> This is not working in new Pandas


```python
df["LotFrontage"] = df["LotFrontage"].fillna(df["LotFrontage"].median())
```

```python
df["LotFrontage"].isna().sum() # HERE WE CAN SEE THAT ALL OF THE NA VALUES FOR LOTFRONTAGE COLUMN IS SET TO MEADIAN VALUE 
```

### Important
mode() returns a Series, not a single value. When you pass a Series to fillna(), pandas tries to align it by index, so the NaN rows don't get matched and nothing is filled.

Fix: extract the scalar with [0]

df["GarageFinish"] = df["GarageFinish"].fillna(df["GarageFinish"].mode()[0])

mode()[0] gives you the most frequent value (e.g., "Unf") as a plain string, and fillna then replaces all NaNs with it

```python
# Now Filling the Categorical columns
df["GarageFinish"] = df["GarageFinish"].fillna(df["GarageFinish"].mode()[0])
```
